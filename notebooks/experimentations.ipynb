```python
large_str = '((not p) and (p) and (not q)and (q) and (not por p)and (not por not q) and (not por q)and (por not q)and (por q) and (not qor q)and (not por por not q)and (not por por q)and (not por not qor q)and (por not qor q)and (not por por not qor q))'
```

```python
cnf_large = NormalForm()
cnf_large.parse_str(large_str)
```

```python
cnf_large
```

```python
cnf_large.satisfy().get_localized_dict(cnf_large)
```

```python
p = PropositionalVariable('p')
q = PropositionalVariable('q')
```

```python
ta = TruthAssignment({p: True, q: True})
```

```python
cnf_large.evaluate(ta)
```

```python
for clause in large_str.strip().split('and'):
    clause_cleaned = (clause
                      .strip()
                      .strip('(')
                      .strip(')')
                      )
    clause_obj = Clause()

    for literal in clause_cleaned.strip().split('or'):
        literal_cleaned = literal.strip()
        #It might happen that a normal form not supported
        #by this parse method happens to be input
        #this will cause incorrect parsing that too silently,
        #I plan to implement error in such a scenario to be thrown.
        literal_id, *literal_negator = literal_cleaned.split(' ')[::-1]
        prop_var = PropositionalVariable(literal_id)
        literal_obj = Literal(prop_var, is_negated=bool(literal_negator))
        print('Literal', literal_obj)
        clause_obj.literals.append(literal_obj)
    print('Clause:', clause_obj)
```
